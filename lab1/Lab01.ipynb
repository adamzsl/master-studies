### Zad 1:

Zaimplementuj funkcję:
$f(x) = \frac{1}{8}x^{4} + 2(x + 0.7)^{3} - \frac{1}{2}x^{2} - 38 x + 1$.



```python
# TUTAJ UMIESC KOD
def f(x):
    f = ( 1/8*x**4 ) + 2*((x+0.7)**3)- (1/2*x**2) - 38*x + 1
    return f
```

### Zad 2:

Oblicz wartość funkcji $f(x)$ w przedziale [-20, 20] z krokiem $\Delta = 1$. Wyniki zapisz na liście.



```python
# TUTAJ UMIESC KOD
X = [x for x in range(-20,21)]
print(X)

res = [f(x) for x in X]
print(res)
```

### Zad 3:

Wyświetl wykres funkcji $f(x)$ korzystając z wyników z zad. 2.

```python
# TUTAJ UMIESC KOD
import matplotlib.pyplot as plt

# X = [1,2,3,4,5]
# Y = [1,2,3,4,5]

plt.figure(figsize=(5,5))
plt.title("Funkcja")
plt.xlabel("Oś x")
plt.ylabel("Oś y")
# plt.xticks([1,2,3,4,5])
# plt.yticks([1,2,3,4,5])
# plt.xlim(5,10)
plt.plot(X,res)
```

### Zad 4:

Korzystając z listy otrzymanej w zad. 2, znajdź takie $x$, dla którego funkcja $f(x)$ przyjmuje najmniejszą wartość. Wyświetl $x$ oraz $f(x)$.

```python
# TUTAJ UMIESC KOD

def min_args(X,Y):
    min_y = 99999
    min_i = 0
    for i,y in enumerate(Y):
        if y < min_y:
            min_y = y
            min_i = i
    return min_i, min_y

i,y = min_args(X,res)
print(f"For x: {X[i]}, f(x): {y} at index {i}")
```

### Zad 5:

Wyznacz inny krok ($\Delta' < \Delta$) z zakresu [-20, 20] i sprawdź czy odnajdziesz mniejszą wartość funkcji $f(x)$ niż w poprzednim zadaniu? Jeżeli tak, to podaj $x$ oraz wartość funkcji $f(x)$.

```python
# TUTAJ UMIESC KOD
import numpy as np
X1 = np.linspace(-20,20,400)
Y1 = [f(x) for x in X1]
i,y = min_args(X1,Y1)
print(i,y)
print(f"For x: {X1[i]}, f(x): {y} at index {i}")
```

### Zad 7:

Wygeneruj $k$ losowych wartości $x$ w przedziale [-20, 20](metoda Monte Carlo), a następnie znajdź minimum funkcji $f(x)$.
Wyświetl wykres funkcji.

```python
# TUTAJ UMIESC KOD
import random

def k_random(k):
    print("k: ",k)
    Xk = [random.uniform(-20, 20) for _ in range(k)]
    Yk = [f(x) for x in Xk]
    i,y = min_args(Xk,Yk)
    print(f"For x: {Xk[i]}, f(x): {y} at index {i}")
    plt.xlim(Xk[i] - 1, Xk[i] + 1)
    plt.ylim(y - 1, y + 1)

    plt.xlabel("Oś x")
    plt.ylabel("Oś y")
    plt.plot(X1,Y1)

k = 1000
k_random(k)
    
```
