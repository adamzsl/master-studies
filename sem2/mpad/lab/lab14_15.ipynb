## Serial connection
_In serial connection, data flow between
“edge” nodes is possible only if the state
of the “middle” node is not known._

In my example we have:

**vaccination → flu → pneumonia**

Therefore, if we don't know whether one has a flu, the information about vaccination can influence our belief about pneumonia.

On the other hand, if we know that the patient has flu, the information about vaccination become useless for changing our belief about pneumonia.

## Diverging connection

_In diverging connection, data flow between
“edge” nodes is possible only if the state
of the “middle” node is not known._

**cough ← COVID → smell_loss**

Therefore, if we do not know whether the patient has COVID, the information about cough can influence our belief about smell loss (and vice versa).

On the other hand, if we know whether the patient has COVID, the information about cough becomes useless for changing our belief about smell loss.

## Converging connection

_In converging connection, data flow
between “edge” nodes is possible only if
the state of the “middle” node is known._

**flu → death ← COVID**

If we do not know whether the patient died, the information about flu does not change our belief about COVID, and the information about COVID does not change our belief about flu.

However, if we know that the patient died, the information about flu can influence our belief about COVID (and vice versa), because both conditions can explain the same outcome (death). In particular, if we already know that the patient died and then learn that the patient had a severe flu, this partly explains the death and can reduce our belief that the patient also had a severe COVID infection.

##### 15.1 You were asked to create a simple network for flu diagnosis. You should include two symptoms: coughing and sneezing. You assume that once presence (or absence) of flu is confirmed, observing one symptom does not influence belief in presence of the other symptom.

The structure that satisfies the task is diverging connection

**coughing ← flu → sneezing**

If we don't know whether one has a flu and we observe coughing, chances of sneezing can rise due to assumption that flu might be the reason.

If we know whether one has a flu, observing coughing doesnt change our belief in sneezing as we already may be expecting it.

##### 15.2 After consulting with a doctor, you learn that the above assumption is not correct. Alter the structure of the network accordingly.

After the consultation, we learned that whether one has flu, observing one symptom gives us information about other. 

**coughing ← flu → sneezing**

**coughing → sneezing**

Now coughing and sneezing are not longer independent given flu. If we know that one has a flu, symptoms of the other can rise due to for example natural response of our immune system to the flu.
