## Zadanie 2

Do niżej zaimplentowanych funkcji napisz dodaj poprawne anotacje typów.


> Wskazówka: Zastanów się, jakie typy danych mogą przyjmować argumenty i zwracać funkcje.

```python
def concatenate(a, b):
    return a + b

def divide(x, y):
    return x / y

def sqrt_data(data):
    return {k: v * 2 for k, v in data.items()}
```

````python
def concatenate(a: Union[int,str,List], b: Union[int,str,List]) -> Union[int,str,List]:
    return a + b

def divide(x: Union[int,float], y: Union[int,float]) -> Union[int,float]:
    return x / y

def sqrt_data(data: Dict) -> Dict:
    return {k: v * 2 for k, v in data.items()}

print(concatenate.__annotations__)
print(divide.__annotations__)
print(sqrt_data.__annotations__)
````
