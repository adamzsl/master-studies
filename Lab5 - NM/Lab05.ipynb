### Zadanie 2
Zaimplementuj klasyfikator najbliższej średniej (NM) z zastosowaniem odległości Euklidesa i wykonaj klasyfikację. Wyświetl wynik klasyfikacji (accuracy).

```python
classes = y_train.unique()
means = np.zeros((len(classes), X_train.shape[1])) # mean of each class for each feature
means.shape, classes
```

```python
for i, c in enumerate(classes):
    means[i] = X_train[y_train == c].mean(axis=0) # mean of each column (across all rows -> axis = 0)
means[:3]
```

```python
def edist(x,y):
    return np.sqrt(np.sum((x-y)**2))
```

```python
means_df = pd.DataFrame(means)
def euclidean(X_test, means_df):
    y_pred = []

    for i, row in X_test.iterrows(): # for each test row
        dist = float('inf')
        closest_class = -1
        for im, rowm in means_df.iterrows(): # for the mean of each class
            edistance = edist(row, rowm) 
            if edistance < dist: # choose the closest mean across all features
                dist = edistance
                closest_class = classes[im] # each row in means are for the class in classes (row 0 -> class 36)
        y_pred.append(closest_class)
        
    return np.array(y_pred)
y_pred = euclidean(X_test, means_df)
y_pred.shape, y_test.shape
```

```python
accuracy = np.sum(y_pred == y_test) / len(y_pred)
print(f"Accuracy: {accuracy:.4f}")
```

### Zadanie 3
Zaimplementuj funkcję, która zwraca macierz kowariancji (*uwaga: biblioteka `numpy` posiada gotową implementację `cov` z którą powinieneś porównać swój wynik*).

\begin{equation*}
C = \frac{1}{n - 1} (X - \bar X)(X - \bar X)^T
\end{equation*}

gdzie:
* $X$ to macierz danych,
* $\bar X$ to wektor średnich wartości cech. 



```python
X_mean = X_train.mean(axis=0)
X_mean
```

```python
X_train.shape, X_mean.shape
```

```python
def cov_matrix(X, X_mean):
    n = X.shape[0]
    X_centered = X - X_mean
    # transpose first matrix not the second one, because pandas and numpy have different default orientations than statistical definition
    return (X_centered.T @ X_centered) / (n-1)
my_cov = cov_matrix(X_train, X_mean)
my_cov.shape
```

```python
my_cov
```

```python
cov = np.cov(X_train.T)
cov_np = pd.DataFrame(cov)
```

```python
is_equal = np.allclose(my_cov, cov_np, rtol=1e-5)
print(f"Matrices are equal: {is_equal}")
```

### Zadanie 4
Zaimplementuj klasyfikator najbliższej średniej (NM) z zastosowaniem odległości Mahalanobisa i wykonaj klasyfikację. Wyświetl wynik klasyfikacji (accuracy).

\begin{equation*}
D_j = \sqrt{ (x - \mu_j)^T S_j^{-1}(x - \mu_j) },
\end{equation*}

gdzie:
* $D_j$ to odległość klasyfikowanej próbki od klasy $j$, 
* $\mu_j$ to wektor średnich wartości cech dla klasy $j$, 
* $S_j^{-1}$ to macierz odwrotna do macierzy kowariancji klasy $j$, 
* a $x$ to klasyfikowana próbka.

> Podpowiedź: Do obliczenia macierzy odwrotnej możesz użyć funkcji `linalg.inv` z biblioteki `numpy`.

> UWAGA: W niniejszym zadaniu możesz zastosować dowolną strukturę kodu (nie musisz trzymać się struktury z poprzedniego zadania), jednak algorytm NM należy zaimplementować samodzielnie – bez użycia gotowych rozwiązań (np. z biblioteki `scikit-learn`).

<span style="text-decoration:underline">Referencje</span>

1. Mahalanobis, P C, _On test and measures of group divergence : theoretical formulae_, Journal and Proceedings of Asiatic Society of Bengal (New Series) Vol. 26, pp. 541-588. 1930. (URL: http://library.isical.ac.in:8080/xmlui/bitstream/handle/10263/1639/029.pdf)
2. McLachlan, Goeffrey J. _Mahalanobis distance_, Resonance, pp. 20-26. 1999. (URL: https://www.ias.ac.in/article/fulltext/reso/004/06/0020-0026)

```python
class_cov_matrices = {}

for c in classes:
    each_class = X_train[y_train == c]
    
    class_mean = each_class.mean(axis=0)
    
    cov_m = cov_matrix(each_class, class_mean)
    
    class_cov_matrices[c] = cov_m
```

```python
class_cov_matrices[36].shape
```

```python
def mdist(x, u, S): # x, u -> (14,)
    x = np.array(x)
    u = np.array(u)
    
    x_ = np.array(x - u) # numpy can treat it as a row vector (1,14)
    
    # (1,14) @ (14, 14) -> (1,14) @ (14, 1) -> 1
    return np.sqrt( x_ @ np.linalg.inv(S) @ x_.T )
```

```python
def mahalanobis(X_test, means_df):
    y_pred_maha = []

    for _, row in X_test.iterrows():
        dist = float('inf')
        closest_class = -1
        
        for im, rowm in means_df.iterrows():
            c = classes[im]
            S = class_cov_matrices[c]
            mdistance = mdist(row, rowm, S) 
            
            if mdistance < dist:
                dist = mdistance
                closest_class = c
                
        y_pred_maha.append(closest_class)
        
    return np.array(y_pred_maha)
    

y_pred_maha = mahalanobis(X_test, means_df)
y_pred_maha.shape, y_test.shape
```

```python
accuracy_m = np.sum(y_pred_maha == y_test) / len(y_test)
print(f"Accuracy: {accuracy_m:.4f}")
```

Without scaling the data

```python
y_pred_e = euclidean(X_test, means_df)
y_pred_m = mahalanobis(X_test, means_df)

acc_e = np.sum(y_pred_e == y_test) / len(y_test)
acc_m = np.sum(y_pred_m == y_test) / len(y_test)

print(f"Euclidean accuracy: {acc_e:.4f}")
print(f"Mahalanobis accuracy: {acc_m:.4f}")
```

Why accuracy decreased in using mahalanobis distance and increased in euclidean?

* Euclidean distance treats all dimensions equally, so standardization helps it by putting everything on the same scale

* Mahalanobis inherently normalizes data through the covariance matrix, therefore scaling destroys valuable information about the different variances between classes

* Some class covariance matrices might become close to singular after standardization
